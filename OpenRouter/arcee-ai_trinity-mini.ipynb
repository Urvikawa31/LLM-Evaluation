# arcee-ai/trinity-mini

### Trinity Mini is an AI language model
- the model has 26 billion parameters
- Only 3 billion are used at one time
- sparse MoE model 
- The model is made of many expert sub-models
- This model featuring 128 experts with 8 active per token
- efficient reasoning over long contexts (131k) with robust function calling and multi-step agent workflows


```python
from dotenv import load_dotenv
from openai import OpenAI
import os
```

```python
load_dotenv()
```

```python
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY")
)
```

```python
response = client.chat.completions.create(
    model = "arcee-ai/trinity-mini:free",
    messages=[

        {
            'role' : 'user',
            'content' : """Your answer will be printed directly in a terminal.

                        Guidelines:
                        - Plain text only
                        - Short paragraphs
                        - No bullet markdown
                        - No special formatting
                        
                        Question:
                        What does market capitalization mean in the stock market?
                        """
        }
    ],
extra_body={"reasoning" : {"enabled" : True}}
)
```

```python
print(response)
```
