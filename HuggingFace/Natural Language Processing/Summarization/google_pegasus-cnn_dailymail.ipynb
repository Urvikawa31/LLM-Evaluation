# PEGASUS CNN/DailyMail

## Model Name
**google/pegasus-cnn_dailymail**

---

## Overview
PEGASUS (Pre-training with Extracted Gap-sentences for Abstractive Summarization) is a **Transformer-based abstractive summarization model** developed by **Google Research**.

This version is fine-tuned on the **CNN/DailyMail** dataset, making it especially effective for **news article summarization** and long-form document condensation.

---

## Task
**Abstractive Text Summarization**

- **Input:** Long text (news articles or documents)
- **Output:** Concise abstractive summary

---

## Training Dataset
- **Dataset:** CNN/DailyMail
- **Content Type:** News articles and summaries
- **Objective:** Generate human-like summaries

---

## Model Architecture
- **Architecture Type:** Encoder–Decoder Transformer
- **Model Family:** PEGASUS
- **Specialization:** Abstractive summarization
- **Language:** English

---

## Why Use This Model?
- Produces fluent, human-readable summaries
- Handles long input documents
- Strong performance on news-style content
- Widely used benchmark model for summarization

---

## Example Use Cases
- News article summarization
- Document and report condensation
- Media monitoring and analysis
- Research paper abstract generation
- Knowledge base summarization


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["HF_TOKEN"],
)
```

```python
result = client.summarization(
    """The tower is 324 metres (1,063 ft) tall, 
    about the same height as an 81-storey building, and the tallest structure in Paris. 
    Its base is square, measuring 125 metres (410 ft) on each side. During its construction, 
    the Eiffel Tower surpassed the Washington Monument to become the tallest man-made structure in the world, 
    a title it held for 41 years until the Chrysler Building in New York City was finished in 1930. 
    It was the first structure to reach a height of 300 metres. Due to the addition of a broadcasting aerial at the top of the tower in 1957, 
    it is now taller than the Chrysler Building by 5.2 metres (17 ft). Excluding transmitters, the Eiffel Tower is the second tallest free-standing structure in France after the Millau Viaduct.""",
    model="google/pegasus-cnn_dailymail",
)
```
