# RoBERTa Large OpenAI Detector

## Model Name
**openai-community/roberta-large-openai-detector**

---

## Overview
`roberta-large-openai-detector` is a **text classification model** built on top of the RoBERTa architecture and published by the OpenAI community.  
This model is designed to help detect text that is likely generated by large language models (LLMs), such as ChatGPT and similar systems.

It classifies input text into categories indicating whether it appears to be AI-generated or human-written.

---

## Task
**Binary Text Classification (AI-Generated vs Human Text)**

- **Input**: A string of text
- **Output**: A label indicating whether the text is likely **AI-generated** or **human-written**

---

## Model Architecture
- **Base Model**: RoBERTa-Large
- **Model Type**: Transformer encoder
- **Purpose**: Classification
- **Strengths**:
  - Good at capturing semantic patterns
  - Can differentiate between typical human writing and AI-generated text

---

## Use Cases

This model is useful for:

- Content moderation and verification
- Detecting AI-generated submissions in education
- Identifying machine-generated marketing content
- Filtering autogenerated reviews or comments
- Integrating as a safety layer in applications requiring authenticity


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider = 'hf-inference',
    api_key=os.environ['HF_TOKEN']
)
```

```python
result = client.text_classification("Artificial intelligence has the potential to transform industries by automating complex tasks and improving decision-making through data-driven insights.",
                                    model = "openai-community/roberta-large-openai-detector"
)

result
```

```python
label_map = {
    "LABEL_1": "AI-GENERATED",
    "LABEL_0": "HUMAN-WRITTEN"
}

best = max(result, key=lambda x: x.score)

formatted_result = [{
    "label": label_map[best.label],
    "score": best.score
}]

print(formatted_result)
```
