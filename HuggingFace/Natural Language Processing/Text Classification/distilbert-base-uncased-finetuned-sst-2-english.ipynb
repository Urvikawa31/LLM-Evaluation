# DistilBERT Base Uncased – Fine-tuned on SST-2

## Model Name
**distilbert/distilbert-base-uncased-finetuned-sst-2-english**

---

## Overview
This model is a **DistilBERT-based text classification model** fine-tuned on the **SST-2 (Stanford Sentiment Treebank)** dataset.  
It is designed for **binary sentiment analysis**, classifying text as **positive** or **negative**.

DistilBERT is a **lighter and faster version of BERT**, making this model suitable for real-time and resource-efficient applications.

---

## Task
**Text Classification (Sentiment Analysis)**

- Input: English text (sentence or paragraph)
- Output: Sentiment label
  - `POSITIVE`
  - `NEGATIVE`

---

## Training Dataset
- **Dataset:** SST-2 (Stanford Sentiment Treebank)
- **Language:** English
- **Labels:** Positive, Negative

---

## Model Architecture
- Base Model: DistilBERT
- Type: Transformer encoder
- Parameters: 67 million
- Case Sensitivity: Uncased (lowercase input)

---

## Example Use Cases
- Product review sentiment analysis
- Movie review classification
- Customer feedback analysis
- Social media sentiment monitoring

```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["HF_TOKEN"]
)
```

```python
result = client.text_classification("This product is terrible and stopped working in two days" ,
                                    model = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"
                                    )
```
