# IBM Granite Embeddings 125M (English)

## Model Name
**ibm-granite/granite-embedding-125m-english**

---

## Overview
IBM Granite Embeddings 125M is a **lightweight text embedding model** developed by **IBM** as part of the Granite model family.  
It is designed to convert **English text into dense numerical vectors (embeddings)** that capture semantic meaning.

The model is optimized for **efficiency, low latency, and enterprise-scale usage**, making it suitable for production systems such as search, retrieval, and RAG pipelines.

---

## Task
**Text Embedding / Semantic Representation**

- **Input:** English text (sentence, paragraph, or document)
- **Output:** Dense embedding vector

---

## Model Architecture
- **Model Type:** Transformer-based encoder
- **Parameters:** ~125 million
- **Language:** English
- **Embedding Type:** Dense vector representation
- **Design Goal:** Fast and cost-efficient semantic embeddings

---

## Why Use Granite Embeddings 125M?
- Lightweight and fast
- Low memory and compute requirements
- Suitable for large-scale indexing
- Enterprise-focused design
- Ideal for real-time and batch embedding workloads

---

## Example Use Cases
- Semantic search
- Document similarity
- RAG (Retrieval-Augmented Generation)
- Text clustering
- Recommendation systems
- Knowledge base indexing

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["HF_TOKEN"],
)
```

```python
result = client.sentence_similarity(
    "That is a happy person",
    [
        "That is a happy dog",
        "That is a very happy person",
        "Today is a sunny day"
    ],
    model="ibm-granite/granite-embedding-125m-english",
)

result
```
