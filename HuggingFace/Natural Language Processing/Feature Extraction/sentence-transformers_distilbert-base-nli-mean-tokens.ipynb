# DistilBERT Base NLI – Mean Tokens (Sentence Transformers)

## Model Name
**sentence-transformers/distilbert-base-nli-mean-tokens**

---

## Overview
`distilbert-base-nli-mean-tokens` is a **sentence embedding model** from the **Sentence-Transformers** library.  
It is based on **DistilBERT**, a lightweight and faster version of BERT, and fine-tuned on **Natural Language Inference (NLI)** datasets.

The model generates **fixed-size vector embeddings** for sentences that capture **semantic meaning**, making it suitable for similarity and retrieval tasks.

---

## Task
**Sentence Embedding / Semantic Similarity**

- **Input:** Sentence or short paragraph
- **Output:** Dense numerical vector (embedding)

---

## Training Data
- **Training Objective:** Natural Language Inference (NLI)
- **Datasets:** SNLI, MultiNLI
- **Learning Goal:** Sentences with similar meaning → closer vectors

---

## Model Architecture
- **Base Model:** DistilBERT (uncased)
- **Pooling Strategy:** Mean pooling over token embeddings
- **Embedding Size:** 768 dimensions
- **Language:** English

---

## Why Use This Model?
- Fast and lightweight
- High-quality sentence embeddings
- Easy integration with Sentence-Transformers
- Strong baseline for semantic tasks
- Lower memory usage compared to BERT-large models

---

## Example Use Cases
- Semantic search
- Sentence similarity
- Clustering text data
- Duplicate question detection
- RAG (Retrieval-Augmented Generation)
- Recommendation systems

```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["HF_TOKEN"],
)
```

```python
result = client.feature_extraction(
    "Today is a sunny day and I will get some ice cream.",
    model="sentence-transformers/distilbert-base-nli-mean-tokens",
)

result
```
