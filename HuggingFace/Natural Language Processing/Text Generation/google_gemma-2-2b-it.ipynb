# Gemma 2 – 2B Instruct

## Model Name
**google/gemma-2-2b-it**

---

## Overview
Gemma 2 – 2B Instruct is a **lightweight instruction-tuned language model** developed by **Google** as part of the **Gemma 2** family.  
It is designed to **follow instructions**, generate helpful responses, and perform common NLP tasks efficiently on **low-to-mid resource systems**.

This model balances **quality, safety, and efficiency**, making it suitable for developers who need a capable model without large compute requirements.

---

## Task
**Instruction-Following Text Generation**

- **Input:** Prompt / instruction
- **Output:** Generated text response

---

## Model Architecture
- **Model Family:** Gemma 2
- **Model Type:** Decoder-only Transformer
- **Parameters:** ~2 billion
- **Tuning:** Instruction-tuned (IT)
- **Language:** English (primarily)

---

## Why Use Gemma 2 – 2B IT?
- Efficient and lightweight
- Strong instruction-following behavior
- Suitable for edge and on-device scenarios
- Good balance of performance and cost
- Easy to integrate for chat and assistant use cases

---

## Example Use Cases
- Chatbots and assistants
- Question answering
- Text summarization
- Content drafting
- Educational and productivity tools
- Prototyping LLM applications


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    api_key=os.environ["HF_TOKEN"],
)
```

```python
completion = client.chat.completions.create(
    model="google/gemma-2-2b-it",
    messages=[
        {
            "role": "user",
            "content": "What is the capital of France?"
        }
    ],
)
```

```python
print(completion.choices[0].message.content)  
```
