# T5 Large

## Model Name
**google-t5/t5-large**

---

## Overview
T5 (Text-to-Text Transfer Transformer) Large is a **general-purpose language model** developed by **Google Research**.  
It treats **every NLP task as a text-to-text problem**, meaning both inputs and outputs are plain text.

The **large** variant provides stronger performance than smaller T5 models and is widely used for tasks such as **summarization, translation, question answering, and text generation**.

---

## Task
**Text-to-Text NLP Tasks**

T5 Large can perform multiple tasks by changing the input prompt:
- Text summarization
- Machine translation
- Question answering
- Text classification
- Text generation

---

## Model Architecture
- **Architecture:** Encoder–Decoder Transformer
- **Model Size:** Large
- **Parameters:** ~770 million
- **Input/Output Format:** Text → Text
- **Language:** Primarily English (multilingual variants exist separately)


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["HF_TOKEN"],
)
```

```python
result = client.translation(
    "Меня зовут Вольфганг и я живу в Берлине",
    model="google-t5/t5-large",
)

result
```
