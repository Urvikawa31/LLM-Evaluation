# BART Large MNLI

## Model Name
**facebook/bart-large-mnli**

---

## Overview
BART Large MNLI is a **natural language inference (NLI)** model developed by **Facebook AI (Meta)**.  
It is based on the **BART Large** architecture and fine-tuned on the **Multi-Genre Natural Language Inference (MNLI)** dataset.

This model is widely used for **zero-shot text classification**, where it can classify text into custom labels **without additional training**.

---

## Task
**Natural Language Inference (NLI)**  
**Zero-Shot Text Classification**

- **Input:** A text sequence and a candidate label (as a hypothesis)
- **Output:** Entailment probability indicating whether the text supports the label

---

## Training Dataset
- **Dataset:** MNLI (Multi-Genre Natural Language Inference)
- **Genres:** Fiction, telephone speech, government, travel, letters, and more
- **Objective:** Determine entailment, contradiction, or neutrality

---

## Model Architecture
- **Base Model:** BART Large
- **Model Type:** Encoder–decoder Transformer
- **Parameters:** ~406 million
- **Language:** English

---

## Why Use This Model?
- Enables zero-shot classification
- No labeled training data required
- Flexible for many NLP tasks
- Strong general-language understanding
- Easy integration with Hugging Face pipelines

---

## Example Use Cases
- Zero-shot topic classification
- Intent detection
- Content moderation
- Sentiment classification (custom labels)
- News and document categorization

```python
import os
import requests
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
API_URL = "https://router.huggingface.co/hf-inference/models/facebook/bart-large-mnli"
headers = {
    "Authorization": f"Bearer {os.environ['HF_TOKEN']}",
}
```

```python
def query(payload):
    response = requests.post(API_URL, headers=headers, json=payload)
    return response.json()
```

```python
output = query({
    "inputs": "Hi, I recently bought a device from your company but it is not working as advertised and I would like to get reimbursed!",
    "parameters": {"candidate_labels": ["refund", "legal", "faq"]},
})
```
