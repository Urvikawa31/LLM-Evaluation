# TAPAS Large – Fine-tuned on WikiTableQuestions

## Model Name
**google/tapas-large-finetuned-wtq**

---

## Overview
TAPAS (Table Parsing and Semantic Understanding) is a **Transformer-based model** developed by **Google**, designed specifically for **question answering over tables**.

This version is the **large variant** of TAPAS, fine-tuned on the **WikiTableQuestions (WTQ)** dataset.  
It enables users to ask **natural language questions** about tabular data and receive accurate answers without writing SQL queries.

---

## Task
**Table Question Answering (Table QA)**

- **Input:** A table + a natural language question
- **Output:** Answer derived from the table (cell values, aggregations, or boolean results)

---

## Training Dataset
- **Dataset:** WikiTableQuestions (WTQ)
- **Source:** Wikipedia tables
- **Focus:** Reasoning and answering questions using table structure and content

---

## Model Architecture
- **Base Model:** TAPAS Large
- **Architecture Type:** Transformer encoder
- **Specialization:** Joint understanding of table structure and text
- **Input Representation:** Table cells + question tokens

---

## What Makes TAPAS Special?
- No SQL required
- Understands table rows, columns, and cells
- Supports aggregation (SUM, COUNT, MAX, etc.)
- Works directly with pandas DataFrames

---

## Example Use Cases
- Question answering over CSV or Excel data
- Business intelligence and analytics
- Interactive data exploration
- Financial or statistical table analysis
- Natural language interfaces to structured data


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["HF_TOKEN"]
)
```

```python
table = {
    "Country": ["India", "China", "USA"],
    "Population (millions)": ["1380","1440","331"]
}

query = "Which country has the highest population?"
```

```python
result = client.table_question_answering(
    model="google/tapas-large-finetuned-wtq",
    table=table,
    query=query
)

result
```
