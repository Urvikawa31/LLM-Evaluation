# Albertina 900M – Portuguese (pt-BR) Encoder

## Model Name
**Albertina 900M (Portuguese / pt-BR Encoder)**

---

## Overview
Albertina 900M is a **large encoder-only language model** designed specifically for the **Portuguese language**, with a strong focus on **Brazilian Portuguese (pt-BR)**.

It is optimized for **text understanding tasks**, such as classification, semantic similarity, and information extraction, rather than text generation.  
With ~900 million parameters, it provides strong performance for **Portuguese NLP workloads**.

---

## Task
**Text Encoding / Language Understanding**

- **Input:** Portuguese text (pt-BR)
- **Output:** Dense vector representations or task-specific predictions

---

## Model Architecture
- **Model Type:** Transformer encoder
- **Parameters:** ~900 million
- **Language:** Portuguese (Brazilian Portuguese focus)
- **Design Goal:** High-quality semantic understanding for pt-BR

---

## Why Use Albertina 900M?
- Trained specifically for Portuguese
- Stronger than multilingual models for pt-BR
- Suitable for enterprise and academic NLP tasks
- Encoder-only → efficient for understanding tasks

---

## Common Use Cases
- Text classification (news, sentiment, topics)
- Semantic search in Portuguese
- Document similarity and clustering
- Named Entity Recognition (NER)
- Information extraction
- Preprocessing for RAG pipelines (retrieval side)
