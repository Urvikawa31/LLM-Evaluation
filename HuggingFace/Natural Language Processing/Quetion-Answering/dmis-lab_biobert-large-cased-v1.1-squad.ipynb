# BioBERT Large Cased – Fine-tuned on SQuAD

## Model Name
**dmis-lab/biobert-large-cased-v1.1-squad**

---

## Overview
BioBERT Large Cased is a **biomedical question answering (QA) model** developed by **DMIS Lab**.  
It is based on **BioBERT Large**, a domain-specific adaptation of BERT trained on large-scale **biomedical literature**, and further **fine-tuned on the SQuAD (Stanford Question Answering Dataset)**.

This model is designed to answer **fact-based questions** by extracting precise answers from a given context, especially effective for **biomedical and scientific text**.

---

## Task
**Extractive Question Answering (QA)**

- **Input:** A question and a context paragraph
- **Output:** A text span from the context that answers the question

---

## Training Details
- **Base Model:** BioBERT Large (cased)
- **Fine-tuning Dataset:** SQuAD v1.1
- **Domain Pretraining:** PubMed abstracts and PMC full-text articles
- **Language:** English

---

## Model Architecture
- **Architecture Type:** Transformer encoder (BERT Large)
- **Parameters:** ~340 million
- **Case Sensitivity:** Cased (preserves capitalization)
- **Specialization:** Biomedical and scientific language understanding

---

## Why Use This Model?
- Strong performance on biomedical QA tasks
- Better domain understanding than general BERT models
- Suitable for research papers and clinical text
- Accurate span-level answer extraction

---

## Example Use Cases
- Question answering over biomedical research articles
- Clinical decision support systems
- Scientific document exploration
- Medical knowledge retrieval
- Academic research assistants

```python
import os
from dotenv import load_dotenv
from huggingface_hub import InferenceClient
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider = "hf-inference",
    api_key = os.environ["HF_TOKEN"]
)
```

```python
answer = client.question_answering(
    question="What is my name?",
    context="My name is Clara and I live in Berkeley.",
    model="dmis-lab/biobert-large-cased-v1.1-squad",
)
answer
```
