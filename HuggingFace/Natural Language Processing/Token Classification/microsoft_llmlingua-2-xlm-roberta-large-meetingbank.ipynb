# LLM-Lingua 2 – XLM-RoBERTa Large (MeetingBank)

## Model Name
**microsoft/llmlingua-2-xlm-roberta-large-meetingbank**

---

## Overview
LLM-Lingua 2 is a **prompt compression and text condensation model** developed by **Microsoft**.  
This version is based on **XLM-RoBERTa Large** and fine-tuned on the **MeetingBank dataset**, making it especially effective for **meeting transcripts, conversations, and long dialogue-style text**.

The model is designed to **compress long text while preserving important information**, helping large language models work more efficiently with limited context windows.

---

## Task
**Text Compression / Prompt Compression**

- **Input:** Long text or conversation (e.g., meeting transcript)
- **Output:** Shortened version that retains key information

---

## Primary Purpose
The model helps:
- Reduce prompt length
- Preserve semantic meaning
- Improve downstream LLM performance
- Lower inference cost and latency

It is commonly used **before** sending text to large LLMs.

---

## Training Dataset
- **Dataset:** MeetingBank
- **Data Type:** Meeting transcripts and conversational text
- **Focus:** Retaining key discussion points, decisions, and actions

---

## Model Architecture
- **Base Model:** XLM-RoBERTa Large
- **Model Type:** Transformer encoder
- **Specialization:** Cross-lingual and long-text understanding
- **Optimization Goal:** High compression with minimal information loss

---

## Why Use This Model?
- Handles long meeting transcripts efficiently
- Improves LLM context utilization
- Reduces token usage and cost
- Works well for conversational and structured dialogue
- Supports multilingual text due to XLM-RoBERTa backbone

---

## Example Use Cases
- Compressing meeting transcripts before summarization
- Reducing long chat histories in chatbots
- Preparing long documents for RAG pipelines
- Token-budget optimization for LLM-based systems
- Enterprise note-taking and meeting analysis


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client  = InferenceClient(
    provider = 'hf-inference',
    api_key = os.environ['HF_TOKEN']
)
```

```python
result = client.token_classification("""Please compress this meeting transcript by identifying the most important words while preserving the key meaning: 
                                    "Yesterday’s design review highlighted critical issues with the UI responsiveness, and we agreed to prioritize fixes in the next sprint. 
                                    The team also discussed risk mitigation strategies and updated the project timeline accordingly.""" ,
                                    model = "microsoft/llmlingua-2-xlm-roberta-large-meetingbank"
)

result
```
