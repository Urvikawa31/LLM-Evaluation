# BERT Base Named Entity Recognition (NER)

## Model Name
**dslim/bert-base-NER**

---

## Overview
`bert-base-NER` is a **named entity recognition (NER) model** based on the **BERT Base (uncased)** architecture.  
It is fine-tuned to identify and classify named entities in English text, such as **persons, organizations, locations, and miscellaneous entities**.

This model is widely used as a **general-purpose NER baseline** for NLP applications.

---

## Task
**Named Entity Recognition (NER)**

- **Input:** Raw English text
- **Output:** Detected entities with labels and confidence scores

---

## Supported Entity Types
The model typically identifies the following entity categories:
- **PER** – Person names
- **ORG** – Organizations
- **LOC** – Locations
- **MISC** – Miscellaneous entities

---

## Model Architecture
- **Base Model:** BERT Base (uncased)
- **Model Type:** Transformer encoder
- **Language:** English
- **Fine-Tuning Dataset:** CoNLL-2003

---

## Why Use This Model?
- Strong general-purpose NER performance
- Easy to integrate using Hugging Face pipelines
- Suitable for news, articles, and general text
- Well-documented and widely adopted

---

## Example Use Cases
- Information extraction from news articles
- Resume and document parsing
- Knowledge graph construction
- Preprocessing text for downstream NLP tasks
- Entity-aware search systems


```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
client = InferenceClient(
    provider = 'hf-inference',
    api_key = os.environ['HF_TOKEN']
)
```

```python
result = client.token_classification(
    "Satya Nadella became the CEO of Microsoft in February 2014 and is based in Redmond, Washington.",
    model="dslim/bert-base-NER",
)
```
